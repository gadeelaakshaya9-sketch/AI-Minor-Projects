```python
# Keep a copy of the original dataset

df_clean = df.copy()

print("Original rows:", len(df_clean))
```

```python
# Clean column names

df_clean.columns = df_clean.columns.str.strip().str.lower()

print(df_clean.columns.tolist())
```

```python
# Clean brand names

df_clean["brand"] = df_clean["brand"].str.strip().str.title()

# Correct spelling mistake
df_clean["brand"] = df_clean["brand"].replace({
    "Toyta": "Toyota"
})

print(df_clean["brand"].unique())
```

```python
# Clean fuel type

df_clean["fuel_type"] = df_clean["fuel_type"].str.strip().str.title()

print(df_clean["fuel_type"].unique())
```

```python
# Clean transmission

df_clean["transmission"] = df_clean["transmission"].str.strip().str.lower()

df_clean["transmission"] = df_clean["transmission"].replace({
    "auto": "automatic"
})

df_clean["transmission"] = df_clean["transmission"].str.title()

print(df_clean["transmission"].unique())
```

```python
# Clean owner type

df_clean["owner_type"] = df_clean["owner_type"].replace({
    "First": "1st",
    "Second": "2nd",
    "Third": "3rd"
})

print(df_clean["owner_type"].unique())
```

```python
# Clean city names

df_clean["city"] = df_clean["city"].str.strip().str.title()

df_clean["city"] = df_clean["city"].replace({
    "Hyd": "Hyderabad",
    "Bengaluru": "Bangalore",
    "New Delhi": "Delhi"
})

print(df_clean["city"].unique())
```

```python
# Clean insurance validity

df_clean["insurance_valid"] = (
    df_clean["insurance_valid"]
    .str.strip()
    .str.title()
)

print(df_clean["insurance_valid"].unique())
```

```python
# Convert year to numeric

df_clean["year"] = pd.to_numeric(
    df_clean["year"],
    errors="coerce"
)

print(df_clean["year"].dtype)
```

```python
# Clean kilometres driven

df_clean["km_driven"] = (
    df_clean["km_driven"]
    .astype(str)
    .str.replace(" km", "", regex=False)
)

df_clean["km_driven"] = pd.to_numeric(
    df_clean["km_driven"],
    errors="coerce"
)

print(df_clean["km_driven"].head())
```

```python
# Negative kilometres are not meaningful

df_clean.loc[df_clean["km_driven"] < 0, "km_driven"] = np.nan
```

```python
# Check missing values after cleaning

print(df_clean.isnull().sum())
```

```python
# Check duplicate rows

print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
# Remove duplicate rows

before_duplicates = len(df_clean)

df_clean = df_clean.drop_duplicates()

after_duplicates = len(df_clean)

print("Rows before removing duplicates:", before_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

```python
# Fill missing categorical values with the mode

df_clean["brand"] = df_clean["brand"].fillna(df_clean["brand"].mode()[0])
df_clean["transmission"] = df_clean["transmission"].fillna(
    df_clean["transmission"].mode()[0]
)
```

```python
# Fill missing numerical values with the median

numeric_columns = [
    "year",
    "price_inr",
    "km_driven",
    "mileage_kmpl",
    "engine_cc"
]

for column in numeric_columns:
    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median()
    )
```

```python
# Check missing values after cleaning

print(df_clean.isnull().sum())
```

```python
# Chart 6: Correlation heatmap

plt.figure(figsize=(10, 7))

correlation = df_clean[numerical_columns].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()

plt.show()
```

```python
# Chart 7: Average price by car brand

average_price_by_brand = (
    df_clean.groupby("brand")["price_inr"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

average_price_by_brand.plot(kind="bar")

plt.title("Average Car Price by Brand")
plt.xlabel("Car Brand")
plt.ylabel("Average Price (INR)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Outlier analysis using the IQR method

outlier_columns = [
    "price_inr",
    "km_driven",
    "mileage_kmpl",
    "engine_cc"
]

for column in outlier_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ]

    print("\nColumn:", column)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Number of outliers:", len(outliers))
```

```python
# Box plot for price outliers

plt.figure(figsize=(10, 4))

sns.boxplot(x=df_clean["price_inr"])

plt.title("Outliers in Car Prices")
plt.xlabel("Price (INR)")

plt.tight_layout()
plt.show()
```

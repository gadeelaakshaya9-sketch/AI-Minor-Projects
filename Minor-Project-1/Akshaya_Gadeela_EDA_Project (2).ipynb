## 5. Check and Remove Duplicate Records
The guidelines require duplicate rows to be identified, removed if present, and compared before and after cleaning.

```python
duplicate_count = df.duplicated().sum()
before_duplicates = len(df)

print("Duplicate rows found:", duplicate_count)

df_clean = df.drop_duplicates().copy()
after_duplicates = len(df_clean)

print("Rows before removing duplicates:", before_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Rows removed:", before_duplicates - after_duplicates)
```

### Observation
The original dataset contains **31,994 duplicate rows**. These duplicate records were removed so that repeated identical observations do not distort frequency counts, averages, or relationships.

## 6. Handle Missing Values
First, missing values are examined in every column.

```python
missing_before = df_clean.isna().sum().sort_values(ascending=False)
missing_before = missing_before[missing_before > 0]
display(missing_before.to_frame("Missing Values"))
```

### Missing-Value Treatment
- `children`: missing values are treated as 0 because the variable represents the number of children in the booking.
- `country`: missing values are replaced with `"Unknown"` because the country is categorical and the missing country should not be invented.
- `agent` and `company`: missing values are replaced with 0, representing no recorded agent/company rather than fabricating an identifier.

```python
df_clean['children'] = df_clean['children'].fillna(0)
df_clean['country'] = df_clean['country'].fillna('Unknown')
df_clean['agent'] = df_clean['agent'].fillna(0)
df_clean['company'] = df_clean['company'].fillna(0)

print("Missing values after treatment:")
remaining_missing = df_clean.isna().sum()
display(remaining_missing[remaining_missing > 0].to_frame("Missing Values"))
```

## 7. Clean the Dataset
The cleaning stage checks data types, text consistency, and invalid values.

```python
# Convert reservation date to datetime
df_clean['reservation_status_date'] = pd.to_datetime(
    df_clean['reservation_status_date'], errors='coerce'
)

# Remove leading/trailing spaces from categorical text columns
text_cols = df_clean.select_dtypes(include='object').columns
for col in text_cols:
    df_clean[col] = df_clean[col].str.strip()

# Standardize month names
month_order = [
    'January','February','March','April','May','June',
    'July','August','September','October','November','December'
]
df_clean['arrival_date_month'] = pd.Categorical(
    df_clean['arrival_date_month'],
    categories=month_order,
    ordered=True
)

# Create useful derived variables
df_clean['total_nights'] = (
    df_clean['stays_in_weekend_nights'] +
    df_clean['stays_in_week_nights']
)
df_clean['total_guests'] = (
    df_clean['adults'] +
    df_clean['children'] +
    df_clean['babies']
)

# Identify invalid records
invalid_adr = (df_clean['adr'] < 0).sum()
zero_guest_records = (df_clean['total_guests'] == 0).sum()

print("Negative ADR records:", invalid_adr)
print("Records with zero total guests:", zero_guest_records)

# Remove records that are clearly invalid for booking analysis
df_clean = df_clean[df_clean['adr'] >= 0].copy()
df_clean = df_clean[df_clean['total_guests'] > 0].copy()

print("Shape after invalid-record handling:", df_clean.shape)
```

### Observation
One record had a negative Average Daily Rate (ADR), which is not meaningful as a room price, so it was removed. Records with zero total guests were also removed because a booking with no adults, children, or babies cannot represent an actual guest stay. No artificial correction was applied to legitimate extreme values such as a high ADR or long lead time.

## 10. Detect and Handle Outliers
Boxplots are used to identify extreme values. Extreme values are not automatically removed because they may represent genuine hotel bookings.

```python
plt.figure(figsize=(9,5))
plt.boxplot(df_clean['adr'].dropna(), vert=False)
plt.title('ADR Boxplot for Outlier Detection')
plt.xlabel('ADR')
plt.tight_layout()
plt.show()

q1 = df_clean['adr'].quantile(0.25)
q3 = df_clean['adr'].quantile(0.75)
iqr = q3 - q1
adr_upper = q3 + 1.5 * iqr
adr_outliers = (df_clean['adr'] > adr_upper).sum()

print(f"ADR upper IQR boundary: {adr_upper:.2f}")
print(f"ADR values above the IQR boundary: {adr_outliers}")
```

### Outlier Decision
The identified high ADR values are retained because a high hotel room rate can be a genuine observation rather than a data-entry error. The project therefore documents the outliers instead of deleting them without justification.

## 12. Bivariate Analysis
Bivariate analysis studies relationships between two variables.
### Lead Time vs Cancellation

```python
plt.figure(figsize=(8,5))
lead_groups = [df_clean.loc[df_clean['is_canceled']==v, 'lead_time'].dropna() for v in [0,1]]
plt.boxplot(lead_groups, labels=['Not Canceled','Canceled'])
plt.title('Lead Time by Cancellation Status')
plt.xlabel('Cancellation Status')
plt.ylabel('Lead Time (days)')
plt.tight_layout()
plt.show()

display(
    df_clean.groupby('is_canceled')['lead_time']
    .agg(['count','mean','median','min','max'])
)
```

### ADR by Hotel Type

### Observation
Comparing lead time by cancellation status helps identify whether advance planning is associated with cancellations. Comparing ADR by hotel type shows how pricing distributions differ between City Hotel and Resort Hotel bookings.

## 13. Multivariate Analysis
Multiple variables are examined together using grouped summaries and correlation analysis.

```python
# Cancellation rate by hotel
hotel_cancel = df_clean.groupby('hotel')['is_canceled'].mean().mul(100).sort_values(ascending=False)
display(hotel_cancel.to_frame("Cancellation Rate (%)"))

plt.figure(figsize=(8,5))
plt.bar(hotel_cancel.index.astype(str), hotel_cancel.values)
plt.title('Cancellation Rate by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Cancellation Rate (%)')
plt.tight_layout()
plt.show()
```

### Correlation Heatmap

```python
corr_cols = [
    'is_canceled', 'lead_time', 'stays_in_weekend_nights',
    'stays_in_week_nights', 'adults', 'children', 'babies',
    'booking_changes', 'days_in_waiting_list', 'adr',
    'required_car_parking_spaces', 'total_of_special_requests',
    'total_nights', 'total_guests'
]

corr = df_clean[corr_cols].corr()

plt.figure(figsize=(12,9))
plt.imshow(corr, cmap='coolwarm', aspect='auto', vmin=-1, vmax=1)
plt.colorbar(label='Correlation')
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.index)), corr.index)
for i in range(len(corr.index)):
    for j in range(len(corr.columns)):
        plt.text(j, i, f'{corr.iloc[i,j]:.2f}', ha='center', va='center', fontsize=7)
plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```

### Observation
The correlation matrix helps identify numerical variables that move together. Stay-related variables are naturally related to total nights, while cancellation and booking variables can be examined for weaker or stronger relationships. Correlation indicates association, not causation.
